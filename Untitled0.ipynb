```python
class circle:
  def dim(self, r):
    self.r= r
    self.pi=3.14
  def area_circle(self):
    return self.pi*self.r**2
  def cir_circle(self):
    return  2*self.pi*self.r
```

```python
c=circle()
c.dim(5)
print(c.area_circle)
print(c.cir_circle)
```
